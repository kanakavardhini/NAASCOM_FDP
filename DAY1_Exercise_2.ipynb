```python
import numpy as np

# Student scores in Math, Science, English

student1 = np.array([85, 90, 88])
student2 = np.array([80, 85, 82])
student3 = np.array([60, 70, 65])

print("Student 1:", student1)
print("Student 2:", student2)
print("Student 3:", student3)

# Norms (overall performance)

norm1 = np.linalg.norm(student1)
norm2 = np.linalg.norm(student2)
norm3 = np.linalg.norm(student3)

print("\nNorm of Student 1:", norm1)
print("Norm of Student 2:", norm2)
print("Norm of Student 3:", norm3)

# Cosine Similarity Function

def cosine_similarity(a, b):
    dot_product = np.dot(a, b)
    norm_a = np.linalg.norm(a)
    norm_b = np.linalg.norm(b)

    return dot_product / (norm_a * norm_b)

# Similarities

sim12 = cosine_similarity(student1, student2)
sim13 = cosine_similarity(student1, student3)
sim23 = cosine_similarity(student2, student3)

print("\nCosine Similarity")
print("Student1 vs Student2 =", sim12)
print("Student1 vs Student3 =", sim13)
print("Student2 vs Student3 =", sim23)
```
