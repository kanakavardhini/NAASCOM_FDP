```python
import numpy as np

# Coefficient matrix
A = np.array([
    [2, 1, 1],
    [1, 3, 2],
    [3, 2, 4]
])

# Result matrix
b = np.array([110, 180, 290])

# Solve the system
x = np.linalg.solve(A, b)

print("Prices of Products:")
print("Product A =", x[0])
print("Product B =", x[1])
print("Product C =", x[2])

# Verify the solution
print("\nVerification:")
print(A @ x)
print("Expected:", b)
```
